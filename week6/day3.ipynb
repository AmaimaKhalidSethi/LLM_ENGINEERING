# Evaluating Model Performance: The $y$ vs. $\hat{y}$ Scatter Plot

A **Predicted vs. Actual plot** (or $y$ vs. $\hat{y}$ plot) is a fundamental visual tool used to evaluate the accuracy of a regression model. It maps the model's generated predictions directly against the true, observed values.

---

## Core Elements of the Plot

* **$y$ (Actual Value):** Represented on one axis, this is the ground truth (e.g., the true market price of an asset).
* **$\hat{y}$ (Predicted Value / "Y-hat"):** Represented on the opposing axis, this is the price estimated by your machine learning model.
* **The Identity Line ($y = \hat{y}$):** The diagonal reference line (often styled in blue) representing a perfect model. At every point on this line, the predicted value exactly matches the actual value.

---

## Interpreting the Visual Data

Because the ultimate goal is for $\hat{y}$ to equal $y$, the proximity of the coordinates to the reference line dictates the model's precision:

* **High-Accuracy Regions:** Points clustered tightly along the diagonal line indicate where the model is highly accurate.
* **Overestimation:** Points falling significantly **above** the identity line indicate that the model predicted a value higher than the actual asset price.
* **Underestimation:** Points falling significantly **below** the identity line indicate that the model predicted a value lower than the actual asset price.



# Statistical Metrics for Regression: MSE and $R^2$

While a $y$ vs. $\hat{y}$ scatter plot provides an excellent visual diagnostic of a model's performance, statistical metrics quantify that performance into concrete numbers. The two most ubiquitous metrics for evaluating regression models are **Mean Squared Error (MSE)** and the **Coefficient of Determination ($R^2$)**.

---

## 1. Mean Squared Error (MSE)

**Mean Squared Error** measures the average squared difference between the estimated values and the actual values. It quantifies the variance of the residuals (errors).

### Mathematical Definition
For a dataset with $n$ observations, MSE is calculated using the following formula:

$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

Where:
* $y_i$ is the actual value for the $i$-th observation.
* $\hat{y}_i$ is the predicted value for the $i$-th observation.
* $n$ is the total number of data points.

### Key Characteristics
* **The Penalty Effect:** Because the errors are squared, larger errors are penalized exponentially more than smaller errors. For example, an error of \$10 becomes a penalty of 100, while an error of \$100 becomes a penalty of 10,000. 
* **Scale Dependency:** MSE is output in the square of the target variable's units (e.g., if you are predicting price in USD, MSE is expressed in $\text{USD}^2$). This can make direct interpretation less intuitive, which is why practitioners often take its square root to get the **Root Mean Squared Error (RMSE)**.
* **Target Optimization:** An MSE of **0** indicates perfect prediction accuracy. The lower the MSE, the better the model.

---

## 2. Coefficient of Determination ($R^2$ Score)

The **$R^2$ score** measures the proportion of variance in the dependent variable that is predictable from the independent variables. It provides an absolute measure of goodness-of-fit relative to a baseline model.



### Mathematical Definition
$R^2$ compares the residual sum of squares ($SS_{res}$) against the total sum of squares ($SS_{tot}$), which represents the variance inherent to the data when simply guessing the baseline mean ($\bar{y}$):

$$R^2 = 1 - \frac{SS_{res}}{SS_{tot}} = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}$$

### Key Interpretation Rules
* **$R^2 = 1$:** The model explains 100% of the variance. Predictions perfectly match actual values (corresponds to all points sitting exactly on the identity line of your scatter plot).
* **$R^2 = 0$:** The model performs no better than a simple baseline horizontal line tracking the mean ($\bar{y}$) of the data. It has learned no meaningful patterns.
* **Negative $R^2$:** The model is performing worse than simply guessing the average value for every data point. This typically indicates a fundamentally flawed model architecture or severe underfitting.

```python
def get_features(item):
    return {
        "weight": item.weight,
        "weight_unknown": 1 if item.weight==0 else 0,
        "text_length": len(item.summary)
    }
```

```python
def list_to_dataframe(items):
    features = [get_features(item) for item in items]
    df = pd.DataFrame(features)
    df['price'] = [item.price for item in items]
    return df

train_df = list_to_dataframe(train)
test_df = list_to_dataframe(test)
```

```python
# Traditional Linear Regression!

np.random.seed(42)

# Separate features and target
feature_columns = ['weight', 'weight_unknown', 'text_length']

X_train = train_df[feature_columns]
y_train = train_df['price']
X_test = test_df[feature_columns]
y_test = test_df['price']

# Train a Linear Regression
model = LinearRegression()
model.fit(X_train, y_train)

for feature, coef in zip(feature_columns, model.coef_):
    print(f"{feature}: {coef}")
print(f"Intercept: {model.intercept_}")

# Predict the test set and evaluate
y_pred = model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse}")
print(f"R-squared Score: {r2}")
```

```python
def linear_regression_pricer(item):
    features = get_features(item)
    features_df = pd.DataFrame([features])
    return model.predict(features_df)[0]
```

```python
evaluate(linear_regression_pricer, test)
```

```python
prices = np.array([float(item.price) for item in train])
documents = [item.summary for item in train]
```

```python
np.random.seed(42)
vectorizer = CountVectorizer(max_features=2000, stop_words='english')
X = vectorizer.fit_transform(documents)
```

```python
# Here are the 1,000 most common words that it picked, not including "stop words":

selected_words = vectorizer.get_feature_names_out()
print(f"Number of selected words: {len(selected_words)}")
print("Selected words:", selected_words[1000:1020])
```

```python
regressor = LinearRegression()
regressor.fit(X, prices)
```

```python
def natural_language_linear_regression_pricer(item):
    x = vectorizer.transform([item.summary])
    return max(regressor.predict(x)[0], 0)
```

```python
evaluate(natural_language_linear_regression_pricer, test)
```

```python
subset = 15_000
rf_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=4)
rf_model.fit(X[:subset], prices[:subset])
```

## Random Forest model

The Random Forest is a type of "**ensemble**" algorithm, meaning that it combines many smaller algorithms to make better predictions.

It uses a very simple kind of machine learning algorithm called a **decision tree**. A decision tree makes predictions by examining the values of features in the input. Like a flow chart with IF statements. Decision trees are very quick and simple, but they tend to overfit.

In our case, the "features" are the elements of the Vector - in other words, it's the number of times that a particular word appears in the product description.

So you can think of it something like this:

**Decision Tree**  
\- IF the word "TV" appears more than 3 times THEN  
-- IF the word "LED" appears more than 2 times THEN  
--- IF the word "HD" appears at least once THEN  
---- Price = $500


With Random Forest, multiple decision trees are created. Each one is trained with a different random subset of the data, and a different random subset of the features. You can see above that we specify 100 trees, which is the default.

Then the Random Forest model simply takes the average of all its trees to product the final result.

```python
def random_forest(item):
    x = vectorizer.transform([item.summary])
    return max(0, rf_model.predict(x)[0])
```

```python
evaluate(random_forest, test)
```

## Introducing XGBoost

Like Random Forest, XGBoost is also an ensemble model that combines multiple decision trees.

But unlike Random Forest, XGBoost builds one tree after another, with each next tree correcting for errors in the prior trees, using 'gradient descent'.

It's much faster than Random Forest, so we can run it for the full dataset, and it's typically better at generalizing.

**If this import doesn't work, please skip this! It's not required. On a Mac, you might need to do `brew install libomp` in the terminal.**

```python
import xgboost as xgb
```

```python
np.random.seed(42)

xgb_model = xgb.XGBRegressor(n_estimators=1000, random_state=42, n_jobs=4, learning_rate=0.1)
xgb_model.fit(X, prices)
```

```python
def xg_boost(item):
    x = vectorizer.transform([item.summary])
    return max(0, xgb_model.predict(x)[0])
```

```python
evaluate(xg_boost, test)
```
